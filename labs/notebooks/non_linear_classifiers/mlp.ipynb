### Exercise 2.2 Implement Backpropagation for an MLP in Numpy and train it
Instantiate the feed-forward model class and optimization parameters. This models follows the architecture described in Algorithm 7.

```python
# Model
geometry = [corpus.nr_features, 20, 20, 2]
activation_functions = ['sigmoid', 'sigmoid', 'softmax']

# Optimization
learning_rate = 0.05
num_epochs = 10
batch_size = 30

import numpy as np
from lxmls.deep_learning.mlp import MLP
from lxmls.deep_learning.utils import index2onehot, logsumexp


class NumpyMLP(MLP):
    """
    Basic MLP with forward-pass and gradient computation in Numpy
    """

    def __init__(self, **config):

        # This will initialize
        # self.config
        # self.parameters
        MLP.__init__(self, **config)

    def predict2(self, input=None):
        """
        Predict model outputs given input
        """
        _,log_class_probabilities = self.log_forward(input)
        return np.argmax(np.exp(log_class_probabilities), axis=1)
    def predict(self, input=None):
        """
        Predict model outputs given input
        """
        log_class_probabilities,_ = self.log_forward(input)
        return np.argmax(np.exp(log_class_probabilities), axis=1)

    def update(self, input=None, output=None):
        """
        Update model parameters given batch of data
        """

        loss,gradients = self.backpropagation(input, output)

        learning_rate = self.config['learning_rate']
        num_parameters = len(self.parameters)
        for m in np.arange(num_parameters):

            # Update weight
            self.parameters[m][0] -= learning_rate * gradients[m][0]

            # Update bias
            self.parameters[m][1] -= learning_rate * gradients[m][1]
        return loss
        
    def log_forward2(self, input):
        """Forward pass for sigmoid hidden layers and output softmax"""
        assert len(self.parameters) == 2
        W_1, b_1 = self.parameters[0]
        W_2, b_2 = self.parameters[1]
        
        z_1 =  input @ W_1.T+ b_1
        z_1_tilde = 1/(1+np.exp(-z_1))
        z_2 = z_1_tilde @ W_2.T +b_2
        log_z_2_tilde = z_2 - np.log(np.sum(np.exp(z_2), axis=1,keepdims=True))

        return z_1_tilde, log_z_2_tilde
        
    def log_forward(self, input):
        """Forward pass for sigmoid hidden layers and output softmax"""
        inputs = [input]
        # hidden layers
        for W, b in self.parameters[:-1]:
            z =  inputs[-1] @ W.T+ b
            z_tilde = 1/(1+np.exp(-z))
            inputs.append(z_tilde)
        
        # output/classif
        W, b = self.parameters[-1]
        z = inputs[-1] @ W.T+ b
        log_z_tilde = z - np.log(np.sum(np.exp(z), axis=1,keepdims=True))

        return log_z_tilde, inputs

    def cross_entropy_loss(self, input, output):
        """Cross entropy loss"""
        num_examples = input.shape[0]
        log_probability,_ = self.log_forward(input)
        return -log_probability[range(num_examples), output].mean()
        
    def cross_entropy_loss2(self, input, output):
        """Cross entropy loss"""
        num_examples = input.shape[0]
        _,log_probability = self.log_forward(input)
        return -log_probability[range(num_examples), output].mean()

    def backpropagation2(self, input, output):
        """Gradients for sigmoid hidden layers and output softmax"""
        batch_size = len(input)
        z_1_tilde, log_z_2_tilde = self.log_forward(input)
        loss = - np.sum(log_z_2_tilde[np.arange(batch_size),output]) / batch_size
        
        z_2_tilde = np.exp(log_z_2_tilde)
        W_1, b_1 = self.parameters[0]
        W_2, b_2 = self.parameters[1]
        I = index2onehot(output, W_2.shape[0])

        # backprop error/gradient from the output
        grad_wrt_z_2 = (I-z_2_tilde)
        grad_wrt_z_1_tilde = grad_wrt_z_2@W_2
        grad_wrt_z_1 = grad_wrt_z_1_tilde * z_1_tilde * (1-z_1_tilde)

        # forward error from input
        grad_W_1 = - (grad_wrt_z_1.T @ input)/batch_size
        grad_W_2 =  - (grad_wrt_z_2.T @z_1_tilde)/batch_size
        grad_b_1 = - np.sum(grad_wrt_z_1,0)/batch_size
        grad_b_2 = - np.sum(grad_wrt_z_2,0)/batch_size
                
        return loss, [(grad_W_1,grad_b_1), (grad_W_2,grad_b_2)]
        
    def backpropagation2b(self, input, output):
        """Gradients for sigmoid hidden layers and output softmax"""
        batch_size = len(input)
        z_1_tilde, log_z_2_tilde = self.log_forward(input)
        loss = - np.sum(log_z_2_tilde[np.arange(batch_size),output]) / batch_size
        
        z_2_tilde = np.exp(log_z_2_tilde)
        W_1, b_1 = self.parameters[0]
        W_2, b_2 = self.parameters[1]
        I = index2onehot(output, W_2.shape[0])

        # classif layer
        grad_wrt_z_2 = (I-z_2_tilde)
        grad_W_2 =  - (grad_wrt_z_2.T @z_1_tilde)/batch_size
        grad_b_2 = - np.sum(grad_wrt_z_2,0)/batch_size
        grad_wrt_z_1_tilde = grad_wrt_z_2@W_2
        
        # hidden layers
        grad_wrt_z_1 = grad_wrt_z_1_tilde * z_1_tilde * (1-z_1_tilde)
        grad_W_1 = - (grad_wrt_z_1.T @ input)/batch_size
        grad_b_1 = - np.sum(grad_wrt_z_1,0)/batch_size
                
        return loss, [(grad_W_1,grad_b_1), (grad_W_2,grad_b_2)]
        
    def backpropagation(self, input, output):
        """Gradients for sigmoid hidden layers and output softmax"""
        batch_size = len(input)
        log_z_tilde, inputs = self.log_forward(input)
        loss = - np.sum(log_z_tilde[np.arange(batch_size),output]) / batch_size
        
        z_tilde = np.exp(log_z_tilde)

        # classif layer
        W, b = self.parameters[-1]
        I = index2onehot(output, W.shape[0])
        grad_wrt_z = (I-z_tilde)

        # hidden layers
        gradients = []
        for n in range(len(self.parameters)-1, -1,-1):
            input = inputs[n]
            W,b = self.parameters[n]
            grad_W =  - (grad_wrt_z.T @input)/batch_size
            grad_b = - np.sum(grad_wrt_z,0)/batch_size
            gradients.append((grad_W, grad_b))
            grad_wrt_previous_z_tilde = grad_wrt_z@W
            grad_wrt_z = grad_wrt_previous_z_tilde * input*(1-input)
                
        return loss, gradients[::-1]

model = NumpyMLP(
    geometry=geometry,
    activation_functions=activation_functions,
    learning_rate=learning_rate
)
```

#### Milestone 2:
After you have ensured that your Backpropagation algorithm is correct, you can train a model with the data we have.

```python
# Get batch iterators for train and test
train_batches = data.batches('train', batch_size=batch_size)
test_set = data.batches('test', batch_size=None)[0]

losses = []
accuracies = []
# Epoch loop
for epoch in range(10):

    # Batch loop
    for batch in train_batches:
        loss = model.update(input=batch['input'], output=batch['output'])
        losses.append(loss)
        #print("\t", loss)

    # Prediction for this epoch
    hat_y = model.predict(input=test_set['input'])

    # Evaluation
    accuracy = 100*np.mean(hat_y == test_set['output'])
    accuracies.append(accuracy)
    # Inform user
    print("Epoch %d: accuracy %2.2f %%" % (epoch+1, accuracy))
```
